```python
# folder = "/content/major_project_25_26/cleaned_crop_data"
# Create output directory structure
input_folder = "/content/major_project_25_26/cleaned_crop_data"

output_base = "/content/drive/MyDrive/MAJORPROJECT"
os.makedirs(output_base, exist_ok=True)



# Define possible seasons
seasons = ["kharif", "rabi", "autumn", "winter", "summer", "whole_year"]
for file_name in os.listdir(input_folder):
    if file_name.endswith(".csv"):
        file_path = os.path.join(input_folder, file_name)
        # Extract crop name from filename (before first space or parenthesis)
    df = pd.read_csv(file_path)
    crop_name = os.path.basename(file_path).split('.')[0].lower()
    # Create a folder for the crop
    crop_folder = os.path.join(output_base, crop_name.split('.')[0])
    os.makedirs(crop_folder, exist_ok=True)

    # Loop through each season and create a corresponding file if yield data exists
    for season in seasons:
        yield_col = f"{crop_name}_{season}_yield_tonne_per_hectare"
        print(yield_col)
        if yield_col in df.columns:
            season_df = df[["state", "district", "year", yield_col]]
            if not season_df.empty:
                season_file = os.path.join(crop_folder, f"{season}.csv")
                season_df.to_csv(season_file, index=False)

    # List generated files
    os.listdir(crop_folder)
```
