```python
import pandas as pd
import numpy as np
```

```python
class_df = pd.read_csv("clean_crop_contribution_data/aggregated_data/kmeans_classes/rice_classified.csv")
```

```python
features_df = pd.read_csv('engineered_climate_features_slight_correlation.csv')
```

```python
features_df
```

```python
# Create a join key in class_df
class_df["location"] = class_df["State"] + "_" + class_df["District"]

# Merge the two dataframes
merged_df = class_df.merge(features_df, on="location", how="inner")

# Optional: drop the helper column if not needed
# merged_df = merged_df.drop(columns=["location"])

print(merged_df.shape)
merged_df
```

```python
# Check districts that did not match
set(class_df["location"]) - set(features_df["location"])
```

```python
print(merged_df["Yield_Class"].value_counts())
print(merged_df["Yield_Class"].value_counts(normalize=True))
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
merged_df["Yield_Class_Encoded"] = le.fit_transform(merged_df["Yield_Class"])

print("\nEncoding mapping:")
print(dict(zip(le.classes_, le.transform(le.classes_))))
```

```python
drop_cols = [
    "State",
    "District",
    "location",
    "Median",
    "Max",
    "Yield_Class"
]

X = merged_df.drop(columns=drop_cols + ["Yield_Class_Encoded"])
y = merged_df["Yield_Class_Encoded"]

print("\nFeature shape:", X.shape)
```

```python
from sklearn.utils.class_weight import compute_class_weight

classes = np.unique(y)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y
)

class_weight_dict = dict(zip(classes, weights))
sample_weights = y.map(class_weight_dict)
```

```python
from sklearn.model_selection import StratifiedKFold

kf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

fold_metrics = []
shap_importance_list = []
```

```python
from sklearn.metrics import accuracy_score, f1_score
import xgboost as xgb
import shap

for fold, (train_idx, val_idx) in enumerate(kf.split(X, y)):

    print(f"\n====================")
    print(f"Fold {fold+1}")
    print(f"====================")

    for col in X.columns:
        if X[col].apply(lambda v: isinstance(v, str)).any():
            print("String values found in:", col)

    X_train = X.iloc[train_idx]
    X_val = X.iloc[val_idx]

    y_train = y.iloc[train_idx]
    y_val = y.iloc[val_idx]

    w_train = sample_weights.iloc[train_idx]

    model = xgb.XGBClassifier(
        n_estimators=400,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric="mlogloss"
    )

    model.fit(
        X_train,
        y_train,
        sample_weight=w_train
    )

    # ---------------- Metrics ----------------
    preds = model.predict(X_val)

    acc = accuracy_score(y_val, preds)
    f1 = f1_score(y_val, preds, average="weighted")

    fold_metrics.append((acc, f1))

    print("Accuracy:", acc)
    print("F1:", f1)


    # ---------------- SHAP ----------------
    explainer = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X_val)

    # Handle both old API (list of arrays) and new API (3D array)
    if isinstance(shap_values, list):
        # Old SHAP: list of (n_samples, n_features) arrays, one per class
        shap_vals = np.mean([np.abs(sv) for sv in shap_values], axis=0)  # (n_samples, n_features)
    elif shap_values.ndim == 3:
        # New SHAP: single 3D array (n_samples, n_features, n_classes)
        shap_vals = np.abs(shap_values).mean(axis=2)  # (n_samples, n_features)
    else:
        # Binary: already (n_samples, n_features)
        shap_vals = np.abs(shap_values)

    shap_mean = shap_vals.mean(axis=0)  # (n_features,)

    shap_importance_list.append(shap_mean)
```

```python
fold_metrics = np.array(fold_metrics)

print("\n====================")
print("FINAL CV METRICS")
print("====================")
print("Accuracy mean:", fold_metrics[:,0].mean())
print("Accuracy std:", fold_metrics[:,0].std())

print("F1 mean:", fold_metrics[:,1].mean())
print("F1 std:", fold_metrics[:,1].std())
```

```python
shap_importance = np.mean(shap_importance_list, axis=0)

print("shap_importance shape:", shap_importance.shape)
print("Number of features:", len(X.columns))

# Flatten to 1D properly based on actual shape
if shap_importance.ndim > 1:
    shap_importance = np.abs(shap_importance).mean(axis=tuple(range(shap_importance.ndim - 1)))

print("shap_importance shape after fix:", shap_importance.shape)

importance_df = pd.DataFrame({
    "feature": X.columns,
    "importance": shap_importance
}).sort_values("importance", ascending=False)

print("\nAll features sorted according to SHAP importance:")
print(importance_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,10))
importance_df.head(20).plot(
    x="feature",
    y="importance",
    kind="barh"
)
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
